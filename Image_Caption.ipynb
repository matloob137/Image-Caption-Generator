```python
# Imports and device setup
import os, json, random, re, time
from collections import Counter
from pathlib import Path

import numpy as np
import torch
import torch.nn as nn
import torch.nn.functional as F
from torch.utils.data import Dataset, DataLoader

from torchvision import models, transforms
from PIL import Image

from nltk.translate.bleu_score import sentence_bleu
import nltk
nltk.data.path.append("/root/nltk_data")  # optional

# Device
device = torch.device("cuda" if torch.cuda.is_available() else "cpu")
print("Using device:", device)

# Seeds
SEED = 42
random.seed(SEED)
np.random.seed(SEED)
torch.manual_seed(SEED)
if torch.cuda.is_available():
    torch.cuda.manual_seed_all(SEED)
```

```python
# === EDIT THESE PATHS to match your dataset ===

CAPTIONS_FILE = "/content/captions.txt"  # path to captions.txt (same format as you provided)
OUT_DIR = "/content/image_caption_output"  # where splits, features, models will be stored
# ==============================================


IMAGE_DIR = "/content/drive/My Drive/Images"
image_files = os.listdir(image_folder)
print(f"Total images: {len(image_files)}")
print("First 10 images:", image_files[:10])

os.makedirs(OUT_DIR, exist_ok=True)
FEATURE_DIR = os.path.join(OUT_DIR, "resnet50_features")
os.makedirs(FEATURE_DIR, exist_ok=True)

# quick sanity
print("Images folder exists:", os.path.isdir(IMAGE_DIR))
print("Captions file exists:", os.path.isfile(CAPTIONS_FILE))
```

```python
# Read captions file
image_captions = {}
with open(CAPTIONS_FILE, "r", encoding="utf-8") as f:
    lines = f.readlines()

# Some caption files include a header 'image,caption' in first line
start_line = 0
if lines and lines[0].strip().lower().startswith("image"):
    start_line = 1

for line in lines[start_line:]:
    line = line.strip()
    if not line:
        continue
    # split at first comma (captions may contain commas)
    parts = line.split(",", 1)
    if len(parts) != 2:
        continue
    img_id, caption = parts[0].strip(), parts[1].strip()
    # Standardize image id to base filename (strip possible path)
    img_id = os.path.basename(img_id)
    image_captions.setdefault(img_id, []).append(caption)

print("Total images with captions:", len(image_captions))
# show sample
for k in list(image_captions.keys())[:3]:
    print(k, "->", image_captions[k][:2])
```

```python
# Create splits (consistent and reproducible)
image_ids = sorted(list(image_captions.keys()))
random.shuffle(image_ids)

n = len(image_ids)
train_ids = image_ids[: int(0.8 * n)]
val_ids   = image_ids[int(0.8 * n) : int(0.9 * n)]
test_ids  = image_ids[int(0.9 * n) :]

split_dir = os.path.join(OUT_DIR, "splits")
os.makedirs(split_dir, exist_ok=True)
with open(os.path.join(split_dir, "train_ids.txt"), "w") as f:
    f.write("\n".join(train_ids))
with open(os.path.join(split_dir, "val_ids.txt"), "w") as f:
    f.write("\n".join(val_ids))
with open(os.path.join(split_dir, "test_ids.txt"), "w") as f:
    f.write("\n".join(test_ids))

print("Saved splits. Train / Val / Test counts:", len(train_ids), len(val_ids), len(test_ids))
```

```python
# Normalization function
def normalize_caption(c):
    c = c.lower().strip()
    # remove characters except letters, digits and spaces
    c = re.sub(r"[^a-z0-9\s]", "", c)
    c = re.sub(r"\s+", " ", c).strip()
    return "<start> " + c + " <end>"

# Build vocab from training captions only
train_captions_norm = []
for img in train_ids:
    for cap in image_captions[img]:
        train_captions_norm.append(normalize_caption(cap))

# token counts
counter = Counter()
for cap in train_captions_norm:
    counter.update(cap.split())

# prune low frequency words (threshold can be changed)
MIN_FREQ = 5
vocab_words = [w for w, cnt in counter.items() if cnt >= MIN_FREQ and w not in ("<start>", "<end>", "<pad>", "<unk>")]

# special tokens
specials = ["<pad>", "<start>", "<end>", "<unk>"]
vocab = specials + vocab_words
word2idx = {w: i for i, w in enumerate(vocab)}
idx2word = {i: w for w, i in word2idx.items()}

vocab_size = len(vocab)
print("Vocab size:", vocab_size)
print("Some sample tokens:", vocab[:20])

# determine max sequence length (95th percentile to avoid too long)
lengths = [len(cap.split()) for cap in train_captions_norm]
max_length = int(np.percentile(lengths, 95))
print("Max length (95th percentile):", max_length)

# helper to convert caption -> indices
def caption_to_seq(caption):
    toks = normalize_caption(caption).split()
    return [word2idx.get(t, word2idx["<unk>"]) for t in toks]
```

```python
# ResNet preprocessing & model (global avg pooled features, 2048-D)
preprocess = transforms.Compose([
    transforms.Resize((224, 224)),
    transforms.ToTensor(),
    transforms.Normalize(mean=[0.485, 0.456, 0.406], std=[0.229, 0.224, 0.225]),
])

resnet = models.resnet50(weights=models.ResNet50_Weights.IMAGENET1K_V1)
resnet = nn.Sequential(*list(resnet.children())[:-1])  # remove classification head
resnet.eval().to(device)

# iterate over images in the split (or all images)
all_images = list(image_captions.keys())
print("Will extract features for", len(all_images), "images")

num_skipped = 0
t0 = time.time()
for i, img_name in enumerate(all_images):
    base, _ = os.path.splitext(img_name)
    out_path = os.path.join(FEATURE_DIR, base + ".npy")
    if os.path.exists(out_path):
        num_skipped += 1
        continue  # skip already computed
    img_path = os.path.join(IMAGE_DIR, img_name)
    if not os.path.exists(img_path):
        # some datasets have mismatches; skip missing images
        continue
    img = Image.open(img_path).convert("RGB")
    inp = preprocess(img).unsqueeze(0).to(device)
    with torch.no_grad():
        feat = resnet(inp).squeeze().cpu().numpy()  # shape (2048,)
    np.save(out_path, feat)
    if (i + 1) % 100 == 0:
        print(f"[{i+1}/{len(all_images)}] extracted features...")

t1 = time.time()
print("Done. time:", round(t1 - t0, 1), "s. skipped:", num_skipped)
```

```python
import torch.nn.utils.rnn as rnn_utils

class CaptionDataset(Dataset):
    """
    Returns:
      feature: torch.float32 (2048,)
      input_seq: list[int] (padded to max_length)  -- tokens from <start> to w_{t-1}
      target_seq: list[int] (padded to max_length) -- tokens from w_1 to <end>
    """
    def __init__(self, id_list, image_captions_map, feature_dir, max_length):
        self.samples = []  # list of (img_id, caption_str)
        self.feature_dir = feature_dir
        self.max_length = max_length
        for img_id in id_list:
            caps = image_captions_map.get(img_id, [])
            for c in caps:
                self.samples.append((img_id, c))
    def __len__(self):
        return len(self.samples)
    def __getitem__(self, idx):
        img_id, caption = self.samples[idx]
        base, _ = os.path.splitext(img_id)
        feat_path = os.path.join(self.feature_dir, base + ".npy")
        feat = np.load(feat_path)
        seq = caption_to_seq(caption)
        # split into input (all but last) and target (all but first)
        input_seq = seq[:-1]
        target_seq = seq[1:]
        # pad/truncate to max_length
        if len(input_seq) > self.max_length:
            input_seq = input_seq[:self.max_length]
            target_seq = target_seq[:self.max_length]
        else:
            pad_len = self.max_length - len(input_seq)
            input_seq = input_seq + [word2idx["<pad>"]] * pad_len
            target_seq = target_seq + [word2idx["<pad>"]] * pad_len
        return torch.tensor(feat, dtype=torch.float32), torch.tensor(input_seq, dtype=torch.long), torch.tensor(target_seq, dtype=torch.long)

# create datasets and loaders
BATCH_SIZE = 32
train_dataset = CaptionDataset(train_ids, image_captions, FEATURE_DIR, max_length)
val_dataset   = CaptionDataset(val_ids, image_captions, FEATURE_DIR, max_length)
test_dataset  = CaptionDataset(test_ids, image_captions, FEATURE_DIR, max_length)

train_loader = DataLoader(train_dataset, batch_size=BATCH_SIZE, shuffle=True, num_workers=2, pin_memory=True)
val_loader   = DataLoader(val_dataset, batch_size=BATCH_SIZE, shuffle=False, num_workers=2, pin_memory=True)

print("Dataset sizes (samples):", len(train_dataset), len(val_dataset), len(test_dataset))
```

```python
class CaptionModel(nn.Module):
    def __init__(self, vocab_size, embed_dim=256, hidden_dim=512, num_layers=1, dropout=0.3, pad_idx=0):
        super().__init__()
        self.vocab_size = vocab_size
        self.embed = nn.Embedding(vocab_size, embed_dim, padding_idx=pad_idx)
        # project 2048 -> hidden_dim (for initial h0)
        self.img_proj = nn.Linear(2048, hidden_dim)
        # optionally project image to c0 also (here init c0 with same)
        # LSTM
        self.lstm = nn.LSTM(embed_dim, hidden_dim, num_layers=num_layers, batch_first=True, dropout=dropout if num_layers > 1 else 0)
        # final classifier
        self.fc = nn.Linear(hidden_dim, vocab_size)
    def forward(self, img_feats, input_seqs):
        """
        img_feats: (B, 2048)
        input_seqs: (B, T)  long
        returns logits: (B, T, vocab_size)
        """
        B = img_feats.size(0)
        # embedding
        emb = self.embed(input_seqs)  # (B, T, embed_dim)
        # init hidden states from image features
        h0 = torch.tanh(self.img_proj(img_feats)).unsqueeze(0)  # (1, B, hidden_dim)
        c0 = torch.zeros_like(h0)  # (1, B, hidden_dim)
        out, _ = self.lstm(emb, (h0, c0))  # (B, T, hidden_dim)
        logits = self.fc(out)  # (B, T, vocab)
        return logits

# instantiate
EMBED_DIM = 256
HIDDEN_DIM = 512
model = CaptionModel(vocab_size=vocab_size, embed_dim=EMBED_DIM, hidden_dim=HIDDEN_DIM, pad_idx=word2idx["<pad>"]).to(device)
print(model)
```

```python
# Training hyperparams
NUM_EPOCHS = 5
LR = 1e-4
PAD_IDX = word2idx["<pad>"]

optimizer = torch.optim.Adam(model.parameters(), lr=LR)
criterion = nn.CrossEntropyLoss(ignore_index=PAD_IDX)  # will flatten targets

best_val_loss = float("inf")
save_path = os.path.join(OUT_DIR, "best_caption_model.pth")

for epoch in range(1, NUM_EPOCHS + 1):
    t0 = time.time()
    # ---- train ----
    model.train()
    train_loss = 0.0
    num_batches = 0
    for feats, inp_seq, tgt_seq in train_loader:
        feats = feats.to(device)
        inp_seq = inp_seq.to(device)
        tgt_seq = tgt_seq.to(device)
        optimizer.zero_grad()
        logits = model(feats, inp_seq)  # (B, T, V)
        loss = criterion(logits.view(-1, logits.size(-1)), tgt_seq.view(-1))
        loss.backward()
        optimizer.step()
        train_loss += loss.item()
        num_batches += 1
    train_loss /= max(1, num_batches)

    # ---- validate ----
    model.eval()
    val_loss = 0.0
    num_batches = 0
    with torch.no_grad():
        for feats, inp_seq, tgt_seq in val_loader:
            feats = feats.to(device)
            inp_seq = inp_seq.to(device)
            tgt_seq = tgt_seq.to(device)
            logits = model(feats, inp_seq)
            loss = criterion(logits.view(-1, logits.size(-1)), tgt_seq.view(-1))
            val_loss += loss.item()
            num_batches += 1
    val_loss /= max(1, num_batches)

    # save best
    if val_loss < best_val_loss:
        best_val_loss = val_loss
        torch.save({
            "model_state": model.state_dict(),
            "word2idx": word2idx,
            "idx2word": idx2word,
            "max_length": max_length,
            "vocab_size": vocab_size,
        }, save_path)
        improved = True
    else:
        improved = False

    print(f"Epoch {epoch}/{NUM_EPOCHS}  train_loss: {train_loss:.4f}  val_loss: {val_loss:.4f}  time: {time.time()-t0:.1f}s  saved:{improved}")
```
